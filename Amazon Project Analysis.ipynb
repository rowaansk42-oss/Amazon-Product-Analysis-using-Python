# Top 10 Most Reviewed Product

```python
top_reviewed = df.nlargest(10, "rating_count")

plt.figure(figsize=(12,6))
plt.barh(top_reviewed["product_name"], top_reviewed["rating_count"])

plt.title("Top 10 Most Reviewed Products")
plt.xlabel("Rating Count")
plt.ylabel("Product")

plt.gca().invert_yaxis()

plt.show()
```

# Insight
Products with the highest reviews usually have higher customer trust.

High review count doesn't always mean the highest rating.

# Category-Wise Average Rating

```python
avg_rating = (
    df.groupby("main_category")["rating"]
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10,5))
avg_rating.plot(kind="bar")

plt.title("Average Rating by Category")
plt.ylabel("Average Rating")
plt.xticks(rotation=45)

plt.show()
```

# Insight
Compare customer satisfaction across categories.

Categories with lower ratings may need quality improvements.

# Category-Wise Average Discount

```python
avg_discount = (
    df.groupby("main_category")["discount_percentage"]
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10,5))
avg_discount.plot(kind="bar")

plt.title("Average Discount by Category")
plt.ylabel("Average Discount (%)")
plt.xticks(rotation=45)

plt.show()
```

# Insight
Identify which categories rely more on discount strategies.

# Coorelation Analysis

# Product Distribution By Category

```python
category_count = df["main_category"].value_counts()

plt.figure(figsize=(8,8))

plt.pie(
    category_count,
    labels=category_count.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Product Distribution by Category")

plt.show()
```

# Discount Range Distribution

```python
bins = [0,20,40,60,80,100]
labels = ["0-20%","21-40%","41-60%","61-80%","81-100%"]

df["Discount_Group"] = pd.cut(
    df["discount_percentage"],
    bins=bins,
    labels=labels
)

discount_group = df["Discount_Group"].value_counts().sort_index()
```

```python
plt.figure(figsize=(8,8))

plt.pie(
    discount_group,
    labels=discount_group.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Products by Discount Range")

plt.show()
```
